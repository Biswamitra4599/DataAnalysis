## 3D Contour Plot — Complete Guide (with Variations)
🔹 What is a 3D Contour Plot?

A 3D contour plot represents a scalar function


z=f(x,y)

using contour lines or filled contours projected in 3D space.

Instead of rendering a full surface, contours show level sets of equal height.

Common Uses

- Optimization & loss landscapes

- Probability density visualization

- Gradient interpretation

- Physics fields & PDEs

- Topography

- ML decision surfaces

### Required Imports (common)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

### VARIATION 1 — Basic 3D Contour Plot

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z)

ax.set_title("Basic 3D Contour Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

### VARIATION 2 — 3D Contour with Levels

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z, levels=20)

ax.set_title("3D Contour with Fixed Levels")
plt.show()
```

### VARIATION 3 — Colored 3D Contour (Colormap)

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

cont = ax.contour(X, Y, Z, levels=30, cmap="viridis")

fig.colorbar(cont, ax=ax, shrink=0.6)

ax.set_title("3D Contour with Colormap")
plt.show()
```

### VARIATION 4 — Filled 3D Contour (contourf)

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

cf = ax.contourf(X, Y, Z, levels=40, cmap="plasma")

fig.colorbar(cf, ax=ax, shrink=0.6)

ax.set_title("Filled 3D Contour Plot")
plt.show()
```

### VARIATION 5 — Contours Projected on XY Plane

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z, zdir='z', offset=Z.min(), cmap='viridis')

ax.set_zlim(Z.min(), Z.max())
ax.set_title("Contour Projection on XY Plane")

plt.show()
```

### VARIATION 6 — Projection on All Three Planes

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# XY plane
ax.contour(X, Y, Z, zdir='z', offset=Z.min(), cmap='coolwarm')

# XZ plane
ax.contour(X, Y, Z, zdir='y', offset=Y.min(), cmap='coolwarm')

# YZ plane
ax.contour(X, Y, Z, zdir='x', offset=X.max(), cmap='coolwarm')

ax.set_title("3D Contour Projections on All Planes")
plt.show()
```

### VARIATION 7 — Surface + Contour Combination

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# Surface
ax.plot_surface(X, Y, Z, cmap="viridis", alpha=0.7)

# Contour at base
ax.contour(X, Y, Z, zdir='z', offset=Z.min(), cmap='inferno')

ax.set_title("Surface + Contour Overlay")
plt.show()
```

### VARIATION 8 — Logarithmic / Exponential Surface Contours

```python
Z_log = np.log(X**2 + Y**2 + 1)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z_log, levels=30, cmap="inferno")

ax.set_title("Log-Scaled Contour Surface")
plt.show()
```

### VARIATION 9 — Optimization Landscape (Loss Contours)

```python
def loss(x, y):
    return (x - 1.5)**2 + (y + 2)**2

X, Y = np.meshgrid(np.linspace(-5, 5, 300),
                   np.linspace(-5, 5, 300))

Z = loss(X, Y)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z, levels=40, cmap="coolwarm")

ax.set_xlabel("w1")
ax.set_ylabel("w2")
ax.set_zlabel("Loss")
ax.set_title("Optimization Loss Contour (3D)")

plt.show()
```

### VARIATION 10 — Animated 3D Contour Plot

```python
from matplotlib import animation

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

X, Y = np.meshgrid(np.linspace(-3, 3, 150),
                   np.linspace(-3, 3, 150))

def update(frame):
    ax.clear()
    Z = np.sin(np.sqrt(X**2 + Y**2) - frame * 0.2)
    ax.contour(X, Y, Z, levels=25, cmap="viridis")
    ax.set_zlim(-1, 1)
    ax.set_title(f"Time = {frame}")

ani = animation.FuncAnimation(fig, update, frames=50)
plt.show()
```

### VARIATION 11 — Multiple Contour Sets (Comparison)

```python
Z1 = np.sin(X) * np.cos(Y)
Z2 = np.cos(X) * np.sin(Y)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

ax.contour(X, Y, Z1, levels=25, cmap="Blues")
ax.contour(X, Y, Z2, levels=25, cmap="Reds")

ax.set_title("Multiple 3D Contour Sets")
plt.show()
```

### VARIATION 12 — High-Resolution Scientific Visualization

```python
x = np.linspace(-4, 4, 400)
y = np.linspace(-4, 4, 400)
X, Y = np.meshgrid(x, y)

Z = np.exp(-(X**2 + Y**2)) * np.cos(3*X) * np.sin(3*Y)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

cont = ax.contourf(X, Y, Z, levels=80, cmap="turbo")
fig.colorbar(cont, ax=ax)

ax.set_title("High-Resolution 3D Contour Field")
plt.show()
```

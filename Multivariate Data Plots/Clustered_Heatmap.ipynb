## Basic Heatmap (Matrix Values)
🔹 When you just want to visualize numbers

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Random matrix
np.random.seed(42)
data = np.random.rand(6, 6)

plt.figure(figsize=(6, 5))
sns.heatmap(data, annot=True, fmt=".2f")
plt.title("Basic Heatmap")
plt.show()
```

## Correlation Heatmap (Most Common Use)
🔹 Core EDA step

```python
import pandas as pd

# Sample multivariate data
df = pd.DataFrame({
    "F1": np.random.normal(0, 1, 200),
    "F2": np.random.normal(1, 1.2, 200),
    "F3": np.random.normal(2, 0.8, 200),
    "F4": np.random.normal(0, 1.5, 200),
})

corr = df.corr()

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    square=True
)
plt.title("Correlation Heatmap")
plt.show()
```

## Masked Correlation Heatmap (Cleaner & Professional)
🔹 Remove duplicate information

```python
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    cmap="coolwarm",
    center=0,
    square=True
)
plt.title("Upper-Triangle Correlation Heatmap")
plt.show()
```

## Heatmap with Hierarchical Clustering
🔹 Reveals feature groups automatically

```python
sns.clustermap(
    corr,
    cmap="coolwarm",
    center=0,
    annot=True,
    figsize=(7, 6)
)
plt.show()
```

## Heatmap for Feature Selection

```python
threshold = 0.8

high_corr = corr.abs() > threshold
sns.heatmap(high_corr, cmap="Reds")
plt.title("Highly Correlated Features")
plt.show()
```

## Heatmap for Model Residuals
🔹 Error pattern detection

```python
residuals = df["F1"] - df["F2"]

error_df = pd.DataFrame({
    "Feature_1": df["F1"],
    "Feature_2": df["F2"],
    "Residual": residuals
})

plt.figure(figsize=(6, 5))
sns.heatmap(error_df.corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Residual Correlation Heatmap")
plt.show()
```

## Annotated Heatmap (Publication Grade)

```python
plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="viridis",
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)
plt.title("Publication-Ready Heatmap")
plt.show()
```

## Heatmap for Time-Series Data

```python
# Fake daily data
dates = pd.date_range("2024-01-01", periods=30)
ts = pd.DataFrame(
    np.random.randn(30, 7),
    index=dates,
    columns=["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
)

plt.figure(figsize=(8, 4))
sns.heatmap(ts.T, cmap="coolwarm", center=0)
plt.title("Time-Series Heatmap")
plt.xlabel("Date")
plt.ylabel("Day")
plt.show()
```

## Interactive Heatmap (Plotly)

```python
import plotly.express as px

fig = px.imshow(
    corr,
    text_auto=".2f",
    color_continuous_scale="RdBu_r"
)
fig.update_layout(title="Interactive Correlation Heatmap")
fig.show()
```

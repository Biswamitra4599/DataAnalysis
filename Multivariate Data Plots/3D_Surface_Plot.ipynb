# 3D Surface Plot — Complete Guide (with Variations)
🔹 1. What is a 3D Surface Plot?

A 3D surface plot visualizes a function of two variables:

z=f(x,y)

It is commonly used in:

- Optimization landscapes

- Loss surfaces in ML

- Physics & engineering fields

- Probability density visualization

- Signal processing

- Mathematical demonstrations

## Required Imports (common for all)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

## VARIATION 1 — Basic 3D Surface Plot

```python
# Create grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Function
Z = np.sin(np.sqrt(X**2 + Y**2))

# Plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(X, Y, Z)

ax.set_title("Basic 3D Surface Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

## VARIATION 2 — Surface with Colormap + Colorbar

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    X, Y, Z,
    cmap="viridis",
    edgecolor="none"
)

fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

ax.set_title("3D Surface with Colormap")
plt.show()
```

## VARIATION 3 — Wireframe + Surface Overlay

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap="coolwarm", alpha=0.8)
ax.plot_wireframe(X, Y, Z, color="black", linewidth=0.3)

ax.set_title("Surface + Wireframe Overlay")
plt.show()
```

## VARIATION 4 — Contour Projection on 3 Planes

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap="plasma", alpha=0.85)

# Contour projections
ax.contour(X, Y, Z, zdir='z', offset=Z.min(), cmap='plasma')
ax.contour(X, Y, Z, zdir='x', offset=X.min(), cmap='plasma')
ax.contour(X, Y, Z, zdir='y', offset=Y.max(), cmap='plasma')

ax.set_title("3D Surface with Contour Projections")
plt.show()
```

## VARIATION 5 — High-Resolution Smooth Surface

```python
x = np.linspace(-3, 3, 300)
y = np.linspace(-3, 3, 300)
X, Y = np.meshgrid(x, y)

Z = np.exp(-(X**2 + Y**2))

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    X, Y, Z,
    cmap="inferno",
    rstride=1,
    cstride=1,
    antialiased=True
)

fig.colorbar(surf, shrink=0.6)
ax.set_title("High-Resolution Gaussian Surface")
plt.show()
```

## VARIATION 6 — Multiple Surfaces in One Plot

```python
Z1 = np.sin(X) * np.cos(Y)
Z2 = np.cos(X) * np.sin(Y)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z1, cmap="viridis", alpha=0.7)
ax.plot_surface(X, Y, Z2, cmap="autumn", alpha=0.6)

ax.set_title("Multiple Surfaces in One Plot")
plt.show()
```

## VARIATION 7 — Mathematical Optimization Surface

```python
def loss(x, y):
    return (x - 2)**2 + (y + 1)**2

X, Y = np.meshgrid(np.linspace(-5, 5, 200),
                   np.linspace(-5, 5, 200))

Z = loss(X, Y)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap="coolwarm")

ax.set_title("Loss Function Surface")
ax.set_xlabel("w1")
ax.set_ylabel("w2")
ax.set_zlabel("Loss")

plt.show()
```

## VARIATION 8 — Parametric 3D Surface

```python
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, 2*np.pi, 100)

U, V = np.meshgrid(u, v)

R = 2
r = 0.6

X = (R + r*np.cos(V)) * np.cos(U)
Y = (R + r*np.cos(V)) * np.sin(U)
Z = r * np.sin(V)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap="Spectral")

ax.set_title("Parametric Surface: Torus")
plt.show()
```

## VARIATION 9 — 3D Surface with Lighting & Shading

```python
from matplotlib.colors import LightSource

ls = LightSource(azdeg=315, altdeg=45)

rgb = ls.shade(Z, cmap=plt.cm.terrain)

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, facecolors=rgb, linewidth=0)

ax.set_title("Surface with Lighting & Shading")
plt.show()
```

## VARIATION 10 — Animated 3D Surface (time-varying)

```python
from matplotlib import animation

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

X, Y = np.meshgrid(np.linspace(-3, 3, 100),
                   np.linspace(-3, 3, 100))

def update(frame):
    ax.clear()
    Z = np.sin(np.sqrt(X**2 + Y**2) - frame / 5)
    ax.plot_surface(X, Y, Z, cmap="viridis")
    ax.set_zlim(-1, 1)
    ax.set_title(f"Time = {frame}")

ani = animation.FuncAnimation(fig, update, frames=60)
plt.show()
```

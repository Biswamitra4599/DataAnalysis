## Basic Pair Plot (Seaborn)
🔹 What it shows

* Scatter plots for every variable pair

* Histogram/KDE on the diagonal

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample data
np.random.seed(42)
df = pd.DataFrame({
    "Feature_1": np.random.normal(0, 1, 200),
    "Feature_2": np.random.normal(1, 1.2, 200),
    "Feature_3": np.random.normal(2, 0.8, 200),
    "Feature_4": np.random.normal(0, 1.5, 200),
})

sns.pairplot(df)
plt.show()
```

## Pair Plot with KDE Diagonal (Cleaner)

```python
sns.pairplot(
    df,
    diag_kind="kde",
    plot_kws={"alpha": 0.6, "s": 30},
)
plt.show()

```

## Pair Plot with Hue (Class Labels)
🔹 Extremely important for ML datasets

```python
df["Class"] = np.random.choice(["A", "B"], size=len(df))

sns.pairplot(
    df,
    hue="Class",
    diag_kind="kde",
    plot_kws={"alpha": 0.6},
)
plt.show()
```

## Pair Plot with Regression Lines

```python
sns.pairplot(
    df,
    kind="reg",
    diag_kind="kde",
    plot_kws={"line_kws": {"linewidth": 1}},
)
plt.show()
```

## Large Dataset Optimization (Sampling)

Pair plots do not scale well.

```python
sns.pairplot(
    df.sample(100),
    diag_kind="kde",
    plot_kws={"alpha": 0.5, "s": 15},
)
plt.show()

```

## Hexbin Pair Plot (Density-Based)

```python
sns.pairplot(
    df,
    kind="hex",
    diag_kind="kde",
)
plt.show()
```

## Pair Plot with Custom Variable Selection


```python
sns.pairplot(
    df,
    vars=["Feature_1", "Feature_2", "Feature_3"],
    diag_kind="kde",
)
plt.show()
```

## Pair Plot using Pandas (Pure Scatter Matrix)

```python
pd.plotting.scatter_matrix(
    df,
    figsize=(8, 8),
    diagonal="kde",
    alpha=0.6
)
plt.show()
```

## Pair Plot for Model Residual Analysis


```python
df_res = df.copy()
df_res["Residual"] = df_res["Feature_1"] - df_res["Feature_2"]

sns.pairplot(
    df_res[["Feature_1", "Feature_2", "Residual"]],
    diag_kind="kde"
)
plt.show()

```

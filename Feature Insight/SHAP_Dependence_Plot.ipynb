```python
# ============================================================
# SHAP Dependence Plot — FULL ONE-CELL SCRIPT
# ============================================================

# (1) Install SHAP (run once if needed)
!pip -q install shap

# (2) Imports
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

np.random.seed(42)

# ------------------------------------------------------------
# (3) Dataset
# ------------------------------------------------------------
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# ------------------------------------------------------------
# (4) Train Model
# ------------------------------------------------------------
model = RandomForestClassifier(
    n_estimators=300,
    max_depth=6,
    random_state=42,
    n_jobs=-1
)
model.fit(X_train, y_train)

# ------------------------------------------------------------
# (5) SHAP Explainer
# ------------------------------------------------------------
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)

# For binary classification:
# shap_values[1] → positive class

# ------------------------------------------------------------
# (6) Basic SHAP Dependence Plot (Auto Interaction)
# ------------------------------------------------------------
shap.dependence_plot(
    "mean radius",
    shap_values[1],
    X_test
)

# ------------------------------------------------------------
# (7) SHAP Dependence Plot with Explicit Interaction Feature
# ------------------------------------------------------------
shap.dependence_plot(
    "mean radius",
    shap_values[1],
    X_test,
    interaction_index="mean concavity"
)

# ------------------------------------------------------------
# (8) SHAP Dependence Plot — No Interaction Coloring
# ------------------------------------------------------------
shap.dependence_plot(
    "mean texture",
    shap_values[1],
    X_test,
    interaction_index=None
)

# ------------------------------------------------------------
# (9) SHAP Dependence for Top Important Features (Loop)
# ------------------------------------------------------------
mean_abs_shap = np.abs(shap_values[1]).mean(axis=0)
top_features = X.columns[np.argsort(mean_abs_shap)[-4:]]

for feat in top_features:
    shap.dependence_plot(
        feat,
        shap_values[1],
        X_test
    )

# ------------------------------------------------------------
# (10) SHAP Dependence — Probability Space (Optional)
# ------------------------------------------------------------
# Convert SHAP values from log-odds to probability shift (approx)
from scipy.special import expit

base = explainer.expected_value[1]
prob_shap = expit(base + shap_values[1]) - expit(base)

shap.dependence_plot(
    "mean radius",
    prob_shap,
    X_test
)
```

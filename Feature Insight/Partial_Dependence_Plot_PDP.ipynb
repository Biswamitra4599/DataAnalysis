## Partial Dependence Plot — What it shows (intuition)

PDP shows the average marginal effect of a feature (or feature pair) on model predictions, holding other features fixed.

### Imports & Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import PartialDependenceDisplay

sns.set(style="whitegrid")
np.random.seed(42)
```

```python
# Dataset
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

### Train Model

```python
model = RandomForestClassifier(
    n_estimators=300,
    max_depth=6,
    random_state=42,
    n_jobs=-1
)
model.fit(X_train, y_train)
```


### Single Feature PDP (Basic)

```python
features = ["mean radius"]

PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=features,
    kind="average",
    grid_resolution=100
)

plt.suptitle("Partial Dependence Plot – Single Feature", fontsize=14)
plt.tight_layout()
plt.show()
```

### Multiple Feature PDPs (Side-by-Side)

```python
features = ["mean radius", "mean texture", "mean concavity"]

PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=features,
    n_cols=3,
    grid_resolution=80
)

plt.suptitle("PDP – Multiple Features", fontsize=14)
plt.tight_layout()
plt.show()
```

### PDP with Individual Conditional Expectation (ICE)

```python
PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=["mean radius"],
    kind="both",        # PDP + ICE
    subsample=100,
    grid_resolution=60
)

plt.suptitle("PDP + ICE Curves", fontsize=14)
plt.tight_layout()
plt.show()
```

### Centered PDP (Removes Baseline Bias)

```python
PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=["mean radius"],
    centered=True
)

plt.suptitle("Centered Partial Dependence Plot", fontsize=14)
plt.tight_layout()
plt.show()
```

### Two-Feature PDP (2D Interaction Surface)

```python
features = [("mean radius", "mean concavity")]

PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=features,
    grid_resolution=50
)

plt.suptitle("2D Partial Dependence Plot (Feature Interaction)", fontsize=14)
plt.tight_layout()
plt.show()
```

### PDP for Top-K Important Features

```python
# Select top features using built-in importance
importances = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

top_features = importances.head(5).index.tolist()

PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=top_features,
    n_cols=3
)

plt.suptitle("PDP – Top 5 Important Features", fontsize=14)
plt.tight_layout()
plt.show()
```

### PDP for Classification – Probability Scale

```python
PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=["mean radius"],
    response_method="predict_proba",
    target=1
)

plt.suptitle("PDP on Predicted Probability (Class = 1)", fontsize=14)
plt.tight_layout()
plt.show()
```

### PDP with Custom Figure Layout

```python
fig, ax = plt.subplots(figsize=(6, 4))

PartialDependenceDisplay.from_estimator(
    model,
    X_train,
    features=["mean smoothness"],
    ax=ax
)

ax.set_title("Custom Styled PDP")
ax.set_ylabel("Partial Dependence")
plt.show()
```

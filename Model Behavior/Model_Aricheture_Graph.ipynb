## Model Architecture Graph — What it Represents

- A model architecture graph visually shows:

- Layers (Dense, Conv, Pooling, etc.)

- Connections (flow of tensors)

- Input / Output shapes

- Depth & width of the network

When architecture graphs are useful

✅ Teaching & demos

✅ Debugging shape mismatches

✅ Explaining design choices

✅ Documentation & papers

### MODEL ARCHITECTURE GRAPH — ALL PRACTICAL METHODS

```python
# Install dependencies (run once)
! pip install pydot graphviz
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import plot_model

# --------------------------------------------------
# Define a Neural Network
# --------------------------------------------------
model = models.Sequential([
    layers.Input(shape=(784,), name="Input"),
    layers.Dense(256, activation="relu", name="Dense_1"),
    layers.Dense(128, activation="relu", name="Dense_2"),
    layers.Dense(10, activation="softmax", name="Output")
])

model.summary()
```

```python
# --------------------------------------------------
# Plot Model Architecture
# --------------------------------------------------
plot_model(
    model,
    show_shapes=True,
    show_layer_names=True,
    dpi=120
)
```

### CNN Architecture Graph (Image Models)

```python
cnn = models.Sequential([
    layers.Input(shape=(28, 28, 1), name="Input_Image"),
    layers.Conv2D(32, (3,3), activation="relu", name="Conv_1"),
    layers.MaxPooling2D(2,2, name="Pool_1"),
    layers.Conv2D(64, (3,3), activation="relu", name="Conv_2"),
    layers.MaxPooling2D(2,2, name="Pool_2"),
    layers.Flatten(name="Flatten"),
    layers.Dense(128, activation="relu", name="Dense"),
    layers.Dense(10, activation="softmax", name="Output")
])

plot_model(
    cnn,
    show_shapes=True,
    show_layer_names=True,
    dpi=120
)
```

### PyTorch — Architecture Graph via torchviz

```python
# Install once
!pip install torchviz
```

```python
import torch
import torch.nn as nn
from torchviz import make_dot

# --------------------------------------------------
# Define PyTorch Model
# --------------------------------------------------
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 256)
        self.fc2 = nn.Linear(256, 128)
        self.fc3 = nn.Linear(128, 10)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        return self.fc3(x)

model = SimpleNN()

# Dummy input
x = torch.randn(1, 784)

# --------------------------------------------------
# Create Architecture Graph
# --------------------------------------------------
dot = make_dot(
    model(x),
    params=dict(model.named_parameters())
)
dot
```

### Custom Architecture Graph (Pedagogical Diagram)
✔ Best for slides & theory classes

```python
import matplotlib.pyplot as plt

layers = [
    ("Input", 3),
    ("Hidden 1", 5),
    ("Hidden 2", 4),
    ("Output", 2)
]

fig, ax = plt.subplots(figsize=(6, 4))

x_spacing = 1.5

for i, (name, neurons) in enumerate(layers):
    y_positions = range(neurons)
    for y in y_positions:
        ax.scatter(i * x_spacing, y, s=300)
    ax.text(i * x_spacing, max(y_positions)+0.6, name, ha="center")

ax.axis("off")
ax.set_title("Neural Network Architecture (Conceptual)")
plt.show()
```

### Architecture Graph + Parameters Table (Audit View)

```python
import pandas as pd

summary_data = []

for layer in model.modules():
    if isinstance(layer, nn.Linear):
        summary_data.append({
            "Layer": layer.__class__.__name__,
            "Input Features": layer.in_features,
            "Output Features": layer.out_features,
            "Parameters": layer.weight.numel() + layer.bias.numel()
        })

pd.DataFrame(summary_data)
```

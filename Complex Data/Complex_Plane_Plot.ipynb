## Complex Plane Plot (Argand Diagram)

### Complex Plane Plot (Large Dataset)

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate more complex data
np.random.seed(42)
n = 500

real_part = np.random.normal(loc=0, scale=2, size=n)
imag_part = np.random.normal(loc=0, scale=2, size=n)

z = real_part + 1j * imag_part

# Plot
plt.figure(figsize=(7, 7))
plt.scatter(
    z.real,
    z.imag,
    c=np.abs(z),       # color by magnitude
    cmap='viridis',
    alpha=0.7,
    s=30
)

# Axes lines
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("Complex Plane Plot (Argand Diagram)")
plt.colorbar(label="Magnitude |z|")
plt.axis('equal')
plt.grid(True)
plt.show()
```

### Example: Eigenvalues in the Complex Plane (Many Points)

```python
import numpy as np
import matplotlib.pyplot as plt

# Random matrix
A = np.random.randn(50, 50)
eigvals = np.linalg.eigvals(A)

plt.figure(figsize=(7, 7))
plt.scatter(
    eigvals.real,
    eigvals.imag,
    c=np.abs(eigvals),
    cmap='plasma',
    s=40
)

plt.axhline(0, color='gray')
plt.axvline(0, color='gray')

plt.xlabel("Real(λ)")
plt.ylabel("Imag(λ)")
plt.title("Eigenvalues in Complex Plane")
plt.colorbar(label="|λ|")
plt.axis('equal')
plt.grid(True)
plt.show()
```

### Optional Enhancements
Color by Phase Instead of Magnitude

```python
plt.scatter(z.real, z.imag, c=np.angle(z), cmap='twilight')
plt.colorbar(label="Phase (radians)")
```

### Density Plot for Very Large Data

```python
plt.hexbin(z.real, z.imag, gridsize=50, cmap='inferno')
plt.colorbar(label="Point Density")
```

### Phase of Complex Numbers (∠z vs Index)

```python
import numpy as np
import matplotlib.pyplot as plt

# Example complex data
z = np.array([1+2j, 2-1j, -1+1j, 3+0.5j, -2-2j])

# Phase (radians)
phase = np.angle(z)

# Plot
plt.figure(figsize=(6, 4))
plt.plot(phase, marker='o')
plt.xlabel("Index")
plt.ylabel("Phase (radians)")
plt.title("Phase Plot of Complex Values")
plt.grid(True)
plt.show()
```

### Phase vs Frequency (FFT Phase Spectrum)

```python
import numpy as np
import matplotlib.pyplot as plt

# Signal
fs = 1000
t = np.linspace(0, 1, fs, endpoint=False)
signal = np.sin(2 * np.pi * 50 * t)

# FFT
fft_vals = np.fft.fft(signal)
freqs = np.fft.fftfreq(len(fft_vals), 1/fs)

# Phase
phase = np.angle(fft_vals)

# Plot (positive frequencies only)
plt.figure(figsize=(6, 4))
plt.plot(freqs[:len(freqs)//2], phase[:len(phase)//2])
plt.xlabel("Frequency (Hz)")
plt.ylabel("Phase (radians)")
plt.title("FFT Phase Spectrum")
plt.grid(True)
plt.show()
```

### Unwrapped Phase Plot (Recommended for Signals)

```python
phase_unwrapped = np.unwrap(np.angle(fft_vals))

plt.figure(figsize=(6, 4))
plt.plot(freqs[:len(freqs)//2], phase_unwrapped[:len(phase_unwrapped)//2])
plt.xlabel("Frequency (Hz)")
plt.ylabel("Unwrapped Phase (radians)")
plt.title("Unwrapped Phase Spectrum")
plt.grid(True)
plt.show()
```

### Phase in Degrees

```python
phase_deg = np.degrees(np.angle(z))

plt.plot(phase_deg, marker='o')
plt.ylabel("Phase (degrees)")
plt.grid(True)
plt.show()
```

### Phase Plot in the Complex Plane (Color-coded)

```python
plt.figure(figsize=(6, 5))
plt.scatter(z.real, z.imag, c=np.angle(z), cmap='twilight')
plt.colorbar(label='Phase (radians)')
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.title("Phase in Complex Plane")
plt.axis('equal')
plt.grid(True)
plt.show()
```

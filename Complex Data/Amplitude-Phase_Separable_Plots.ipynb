### Amplitude–Phase Separable Plots (Complex Data)

```python
import numpy as np
import matplotlib.pyplot as plt

# Example complex signal
n = 500
t = np.linspace(0, 1, n)

signal = (
    2 * np.exp(1j * (2 * np.pi * 5 * t)) +
    0.5 * np.exp(1j * (2 * np.pi * 12 * t))
)

# Amplitude and phase
amplitude = np.abs(signal)
phase = np.unwrap(np.angle(signal))

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

# Amplitude plot
ax1.plot(t, amplitude, color='blue')
ax1.set_ylabel("Amplitude")
ax1.set_title("Amplitude Plot")
ax1.grid(True)

# Phase plot
ax2.plot(t, phase, color='red')
ax2.set_xlabel("Time")
ax2.set_ylabel("Phase (radians)")
ax2.set_title("Phase Plot")
ax2.grid(True)

plt.tight_layout()
plt.show()
```

### Frequency-Domain Version (FFT Amplitude & Phase)

```python
# FFT
fft_vals = np.fft.fft(signal)
freqs = np.fft.fftfreq(len(fft_vals), t[1] - t[0])

amplitude = np.abs(fft_vals)
phase = np.unwrap(np.angle(fft_vals))

# Plot (positive frequencies only)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

ax1.plot(freqs[:n//2], amplitude[:n//2])
ax1.set_ylabel("Amplitude")
ax1.set_title("FFT Amplitude Spectrum")
ax1.grid(True)

ax2.plot(freqs[:n//2], phase[:n//2])
ax2.set_xlabel("Frequency (Hz)")
ax2.set_ylabel("Phase (radians)")
ax2.set_title("FFT Phase Spectrum")
ax2.grid(True)

plt.tight_layout()
plt.show()
```

### Basic Phasor Plot

```python
import numpy as np
import matplotlib.pyplot as plt

# Define phasors (magnitude * exp(j*phase))
phasors = {
    "V1": 5 * np.exp(1j * np.deg2rad(30)),
    "V2": 3 * np.exp(1j * np.deg2rad(-45)),
    "V3": 4 * np.exp(1j * np.deg2rad(90))
}

plt.figure(figsize=(7, 7))

for label, z in phasors.items():
    plt.arrow(
        0, 0,
        z.real, z.imag,
        head_width=0.15,
        length_includes_head=True,
        alpha=0.8
    )
    plt.text(
        z.real * 1.05,
        z.imag * 1.05,
        label,
        fontsize=10
    )

# Reference axes
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("Phasor Plot")
plt.axis('equal')
plt.grid(True)
plt.show()
```

### Multiple Phasors from Data (Time-Varying Example)

```python

# Generate phasors from a signal
t = np.linspace(0, 1, 50)
phasors = np.exp(1j * 2 * np.pi * t)

plt.figure(figsize=(7, 7))
plt.quiver(
    np.zeros_like(phasors.real),
    np.zeros_like(phasors.imag),
    phasors.real,
    phasors.imag,
    angles='xy',
    scale_units='xy',
    scale=1
)

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.axis('equal')
plt.title("Rotating Phasors")
plt.grid(True)
plt.show()

```

### Phasor Plot with Magnitude & Phase Annotation

```python
z = 5 * np.exp(1j * np.deg2rad(45))

plt.figure(figsize=(6, 6))
plt.arrow(0, 0, z.real, z.imag, head_width=0.2)

mag = np.abs(z)
phase = np.degrees(np.angle(z))

plt.text(z.real * 0.6, z.imag * 0.6,
         f"|Z|={mag:.2f}\n∠Z={phase:.1f}°")

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.axis('equal')
plt.title("Annotated Phasor")
plt.grid(True)
plt.show()
```

### Magnitude of Complex Numbers (|z| vs Index)

```python
import numpy as np
import matplotlib.pyplot as plt

# Example complex data
z = np.array([1+2j, 2-1j, -1+1j, 3+0.5j, -2-2j])

# Magnitude
magnitude = np.abs(z)

# Plot
plt.figure(figsize=(6, 4))
plt.plot(magnitude, marker='o')
plt.xlabel("Index")
plt.ylabel("|z| (Magnitude)")
plt.title("Magnitude Plot of Complex Values")
plt.grid(True)
plt.show()
```

### Magnitude vs Frequency (FFT Magnitude Spectrum)

```python
import numpy as np
import matplotlib.pyplot as plt

# Signal
fs = 1000  # Sampling frequency (Hz)
t = np.linspace(0, 1, fs, endpoint=False)
signal = np.sin(2 * np.pi * 50 * t) + 0.5 * np.sin(2 * np.pi * 120 * t)

# FFT
fft_vals = np.fft.fft(signal)
freqs = np.fft.fftfreq(len(fft_vals), 1/fs)

# Magnitude
magnitude = np.abs(fft_vals)

# Plot (positive frequencies only)
plt.figure(figsize=(6, 4))
plt.plot(freqs[:len(freqs)//2], magnitude[:len(magnitude)//2])
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Magnitude Spectrum")
plt.grid(True)
plt.show()
```

### Magnitude in dB (Log Scale – Very Common)

```python
magnitude_db = 20 * np.log10(np.abs(fft_vals) + 1e-12)

plt.figure(figsize=(6, 4))
plt.plot(freqs[:len(freqs)//2], magnitude_db[:len(magnitude_db)//2])
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.title("Magnitude Spectrum (dB)")
plt.grid(True)
plt.show()
```

### Scatter Magnitude Plot (Magnitude vs Real/Imag)

```python
plt.figure(figsize=(6, 5))
plt.scatter(z.real, z.imag, c=np.abs(z), cmap='viridis')
plt.colorbar(label='|z|')
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.title("Complex Magnitude Scatter Plot")
plt.axis('equal')
plt.grid(True)
plt.show()
```

### Basic Real vs Imag Scatter Plot

```python
import numpy as np
import matplotlib.pyplot as plt

# Example complex data
z = np.array([1+2j, 2+1j, 3-1j, -1-2j, -2+1j])

# Extract real and imaginary parts
real_part = z.real
imag_part = z.imag

# Plot
plt.figure(figsize=(6, 5))
plt.scatter(real_part, imag_part, color='blue')
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)

plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.title("Real vs Imaginary Scatter Plot")
plt.grid(True)
plt.show()
```

### Example: Eigenvalues (Very Common Use Case)

```python
import numpy as np
import matplotlib.pyplot as plt

# Random matrix
A = np.random.randn(10, 10)

# Eigenvalues
eigvals = np.linalg.eigvals(A)

# Plot
plt.figure(figsize=(6, 5))
plt.scatter(eigvals.real, eigvals.imag, c='red')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("Real(λ)")
plt.ylabel("Imag(λ)")
plt.title("Eigenvalue Spectrum")
plt.grid(True)
plt.show()
```

### Enhancements (Optional)

- Equal axis scaling (recommended):

```python
plt.figure(figsize=(6, 5))
plt.scatter(eigvals.real, eigvals.imag, c='red')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.axis('equal')
plt.xlabel("Real(λ)")
plt.ylabel("Imag(λ)")
plt.title("Eigenvalue Spectrum")
plt.grid(True)
plt.show()


```

- Color by magnitude:

```python
plt.scatter(real_part, imag_part, c=np.abs(z), cmap='viridis')
plt.colorbar(label='|z|')
```

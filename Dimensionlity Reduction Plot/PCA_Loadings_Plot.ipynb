### PCA Loadings Plot (PC1 vs PC2)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Example data (replace with your dataset)
X = np.random.rand(100, 5)
feature_names = [f'Var{i+1}' for i in range(X.shape[1])]

# Standardize data
X_scaled = StandardScaler().fit_transform(X)

# PCA
pca = PCA(n_components=2)
pca.fit(X_scaled)

# Loadings
loadings = pca.components_.T

# Plot
plt.figure(figsize=(8, 8))
plt.scatter(loadings[:, 0], loadings[:, 1])

for i, feature in enumerate(feature_names):
    plt.text(
        loadings[i, 0],
        loadings[i, 1],
        feature,
        ha='center',
        va='center'
    )

plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.xlabel('PC1 Loadings')
plt.ylabel('PC2 Loadings')
plt.title('PCA Loadings Plot')
plt.grid(True)
plt.show()
```

### Optional: Loadings Scaled by Explained Variance (Common in Statistics)

```python
explained_var = np.sqrt(pca.explained_variance_)

scaled_loadings = loadings * explained_var

plt.figure(figsize=(8, 8))
plt.scatter(scaled_loadings[:, 0], scaled_loadings[:, 1])

for i, feature in enumerate(feature_names):
    plt.text(
        scaled_loadings[i, 0],
        scaled_loadings[i, 1],
        feature,
        ha='center',
        va='center'
    )

plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Scaled PCA Loadings Plot')
plt.grid(True)
plt.show()
```

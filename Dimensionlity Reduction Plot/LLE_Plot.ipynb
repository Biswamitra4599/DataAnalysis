## LLE

### Basic LLE Example

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.manifold import LocallyLinearEmbedding

# Load sample data
X, y = load_iris(return_X_y=True)

# Fit LLE
lle = LocallyLinearEmbedding(
    n_neighbors=10,
    n_components=2,
    method='standard',   # 'standard', 'modified', 'hessian', 'ltsa'
    random_state=42
)
X_lle = lle.fit_transform(X)

# Plot
plt.figure(figsize=(6, 5))
plt.scatter(X_lle[:, 0], X_lle[:, 1], c=y, cmap='viridis')
plt.colorbar(label='Class')
plt.title("LLE Plot (Iris Dataset)")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.grid(True)
plt.show()
```

### 3D LLE Plot (Optional)

```python
from mpl_toolkits.mplot3d import Axes3D

lle = LocallyLinearEmbedding(n_neighbors=12, n_components=3)
X_lle_3d = lle.fit_transform(X)

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(
    X_lle_3d[:, 0],
    X_lle_3d[:, 1],
    X_lle_3d[:, 2],
    c=y,
    cmap='viridis'
)
ax.set_title("3D LLE Plot")
plt.show()

```

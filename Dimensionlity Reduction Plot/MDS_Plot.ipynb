### Basic MDS Example (from a distance matrix)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import MDS

# Example distance matrix (must be square)
distance_matrix = np.array([
    [0.0, 1.0, 2.0, 3.0],
    [1.0, 0.0, 1.5, 2.5],
    [2.0, 1.5, 0.0, 1.0],
    [3.0, 2.5, 1.0, 0.0]
])

# Fit MDS
mds = MDS(
    n_components=2,
    dissimilarity="precomputed",
    random_state=42
)
coords = mds.fit_transform(distance_matrix)

# Plot
plt.figure(figsize=(6, 5))
plt.scatter(coords[:, 0], coords[:, 1])

for i in range(len(coords)):
    plt.text(coords[i, 0], coords[i, 1], f"P{i}", fontsize=10)

plt.title("MDS Plot")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```

### MDS Directly from Feature Data (no distance matrix needed)

```python
from sklearn.datasets import load_iris
from sklearn.manifold import MDS
import matplotlib.pyplot as plt

# Load example data
X, y = load_iris(return_X_y=True)

# Fit MDS
mds = MDS(n_components=2, random_state=42)
X_mds = mds.fit_transform(X)

# Plot
plt.figure(figsize=(6, 5))
plt.scatter(X_mds[:, 0], X_mds[:, 1], c=y, cmap="viridis")
plt.colorbar(label="Class")
plt.title("MDS Plot (Iris Dataset)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.show()
```

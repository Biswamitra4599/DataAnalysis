### UMAP 2D Scatter Plot

```python
import numpy as np
import matplotlib.pyplot as plt
import umap
from sklearn.preprocessing import StandardScaler

# Example data (replace with your dataset)
X = np.random.rand(200, 10)
y = np.random.randint(0, 3, size=200)  # optional labels

# Standardize
X_scaled = StandardScaler().fit_transform(X)

# UMAP 2D
umap_2d = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap_2d = umap_2d.fit_transform(X_scaled)

# Plot
plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    X_umap_2d[:, 0],
    X_umap_2d[:, 1],
    c=y,
    cmap='tab10',
    alpha=0.7
)
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')
plt.title('UMAP 2D Scatter Plot')
plt.colorbar(scatter, label='Class')
plt.grid(True)
plt.show()
```

### UMAP 3D Scatter Plot

```python
from mpl_toolkits.mplot3d import Axes3D

# UMAP 3D
umap_3d = umap.UMAP(
    n_components=3,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)

X_umap_3d = umap_3d.fit_transform(X_scaled)

# Plot
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    X_umap_3d[:, 0],
    X_umap_3d[:, 1],
    X_umap_3d[:, 2],
    c=y,
    cmap='tab10',
    alpha=0.7
)

ax.set_xlabel('UMAP 1')
ax.set_ylabel('UMAP 2')
ax.set_zlabel('UMAP 3')
ax.set_title('UMAP 3D Scatter Plot')
fig.colorbar(scatter, ax=ax, label='Class')
plt.show()
```

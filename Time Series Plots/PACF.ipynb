# PACF (Partial Autocorrelation Function) – Python Code

✔ Uses statsmodels

✔ Works for any dataset (list or Series)

✔ Blue bands = 95% confidence

✔ ywm method = most stable estimator

## Simple PACF Plot (Recommended)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_pacf

# ------------------------------
# Sample data (replace with your data)
# ------------------------------
np.random.seed(42)
data = np.cumsum(np.random.randn(200))

series = pd.Series(data)

# ------------------------------
# PACF Plot
# ------------------------------
plt.figure(figsize=(8, 5))
plot_pacf(series, lags=40, method='ywm')
plt.title("Partial Autocorrelation Function (PACF)", fontsize=16)
plt.show()
```

## PACF Using Your Own List

If you have values like:

```python
values = [12, 15, 14, 18, 20, 23, 25, 22]
```

```python
from statsmodels.graphics.tsaplots import plot_pacf
import pandas as pd
import matplotlib.pyplot as plt

values = [12, 15, 14, 18, 20, 23, 25, 22]

series = pd.Series(values)

plt.figure(figsize=(8, 5))
plot_pacf(series, lags=len(values)//2, method='ywm')
plt.title("PACF Plot")
plt.show()
```

## PACF After Differencing (if data has trend)

(This is optional but often necessary.)

```python
series_diff = series.diff().dropna()

lags = len(series_diff) // 2   # valid maximum lag

plt.figure(figsize=(8, 5))
plot_pacf(series_diff, lags=lags, method='ywm')
plt.title("PACF Plot")
plt.show()
```

## What is a Step Plot?

A step plot is used when data changes at specific points (like stock price jumps, sensor signals, digital data, piecewise functions, round-off data, etc.)

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
x = np.arange(1, 11)
y = np.array([3, 5, 4, 8, 6, 7, 9, 8, 10, 12])
```

```python
plt.step(x, y, where='pre')   # step before change
plt.step(x, y, where='mid')   # step centered
plt.step(x, y, where='post')  # step after change
```

### Basic Step Plot (pre, mid, post)

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
x = np.arange(1, 11)
y = np.array([3, 5, 4, 8, 6, 7, 9, 8, 10, 12])

plt.figure(figsize=(12, 5))

plt.step(x, y, where='pre', label='where=pre')
plt.step(x, y, where='mid', label='where=mid')
plt.step(x, y, where='post', label='where=post')

plt.title("Basic Step Plot (pre, mid, post)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

### Step Plot with Markers

```python
plt.figure(figsize=(10,5))
plt.step(x, y, where='mid', linewidth=2)
plt.plot(x, y, 'o', color='red')  # markers at actual data points

plt.title("Step Plot with Markers")
plt.grid(True)
plt.show()
```

### Vertical + Horizontal Step Plot (Staircase)

```python
plt.figure(figsize=(10,5))
plt.step(x, y, where='post', color='black')
plt.title("Staircase Step Plot")
plt.grid(True)
plt.show()
```

### Filled Step Plot (Area Style)

```python
plt.figure(figsize=(10,5))
plt.step(x, y, where='post', color='blue')
plt.fill_between(x, y, step='post', alpha=0.3, color='blue')

plt.title("Filled Step Plot")
plt.grid(True)
plt.show()
```

### Multiple Step Plots on Same Chart

```python
y2 = y + np.random.randint(-2, 3, size=len(y))

plt.figure(figsize=(12,5))
plt.step(x, y, where='post', label='Series 1')
plt.step(x, y2, where='pre', label='Series 2')

plt.title("Multiple Step Plots")
plt.legend()
plt.grid(True)
plt.show()
```

### Step Plot for Time Series

```python
import pandas as pd

dates = pd.date_range("2024-01-01", periods=10)
values = np.random.randint(10, 50, 10)

plt.figure(figsize=(12,5))
plt.step(dates, values, where='post')
plt.plot(dates, values, 'o')

plt.title("Time-Series Step Plot")
plt.grid(True)
plt.show()
```

### Step Plot with Thresholds (Highlight Conditions)

```python
threshold = 7

plt.figure(figsize=(12,5))
plt.step(x, y, where='post', linewidth=2)

# highlight above threshold
plt.fill_between(x, y, threshold, step='post', 
                 where=(y > threshold), alpha=0.3, color='red')

plt.title("Step Plot with Threshold Highlight")
plt.axhline(threshold, color='black', linestyle='--')
plt.grid(True)
plt.show()
```

### Staircase Plot with Step Changes Only

```python
plt.figure(figsize=(12,5))
plt.step(x, y, where='mid', linewidth=2)
plt.scatter(x, y, s=80, color='orange')

plt.title("Staircase Plot with Change Points")
plt.grid(True)
plt.show()
```

### Step Plot with Subplots (Variation Comparison)

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 10))

axes[0].step(x, y, where='pre')
axes[0].set_title("where=pre")

axes[1].step(x, y, where='mid')
axes[1].set_title("where=mid")

axes[2].step(x, y, where='post')
axes[2].set_title("where=post")

plt.tight_layout()
plt.show()
```

### Step Plot + Smoothing Comparison

```python
y_smooth = np.convolve(y, np.ones(3)/3, mode='same')

plt.figure(figsize=(12,5))
plt.step(x, y, where='post', label='Step Original')
plt.plot(x, y_smooth, label='Smoothed Line', linestyle='--')

plt.title("Step Plot vs Smoothed Curve")
plt.legend()
plt.grid(True)
plt.show()
```

### Step Plot with Annotations

```python
plt.figure(figsize=(12,5))
plt.step(x, y, where='post')

for i, v in enumerate(y):
    plt.text(x[i], v+0.5, str(v), fontsize=9)

plt.title("Annotated Step Plot")
plt.grid(True)
plt.show()
```

### Step Plot with Seaborn Styling

```python
import seaborn as sns
sns.set_style("whitegrid")

plt.figure(figsize=(12,5))
plt.step(x, y, where='mid', color='darkblue', linewidth=2)
plt.title("Seaborn Styled Step Plot")
plt.show()
```

## Classic Raincloud Plot (Matplotlib + Seaborn)
🔹 Structure

Half violin → distribution

Boxplot → median & IQR

Strip plot → individual observations/

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data
np.random.seed(42)
data = pd.DataFrame({
    "Value": np.concatenate([
        np.random.normal(0, 1, 300),
        np.random.normal(2, 1.2, 300)
    ]),
    "Group": ["A"] * 300 + ["B"] * 300
})

plt.figure(figsize=(8, 5))

# Half violin
sns.violinplot(
    data=data,
    x="Group",
    y="Value",
    inner=None,
    cut=0,
    linewidth=1,
)

# Boxplot
sns.boxplot(
    data=data,
    x="Group",
    y="Value",
    width=0.15,
    showcaps=True,
    boxprops={"facecolor": "white", "zorder": 3},
    showfliers=False,
)

# Rain (jittered points)
sns.stripplot(
    data=data,
    x="Group",
    y="Value",
    jitter=0.25,
    alpha=0.4,
    size=3,
)

plt.title("Raincloud Plot", fontsize=14)
plt.show()
```

## True Half-Violin Raincloud (More Accurate)

Seaborn draws full violins, but we can clip them to get true half-violins.

```python
plt.figure(figsize=(8, 5))

ax = sns.violinplot(
    data=data,
    x="Group",
    y="Value",
    inner=None,
    cut=0,
)

# Convert to half violins
for violin in ax.collections:
    violin.set_clip_path(
        plt.Rectangle(
            (violin.get_paths()[0].vertices[:, 0].mean(), -10),
            10, 20,
            transform=ax.transData
        )
    )

sns.boxplot(
    data=data,
    x="Group",
    y="Value",
    width=0.15,
    showfliers=False,
)

sns.stripplot(
    data=data,
    x="Group",
    y="Value",
    jitter=0.25,
    alpha=0.4,
)

plt.title("Half-Violin Raincloud Plot")
plt.show()
```

## Horizontal Raincloud Plot (Very Popular)

Used heavily in neuroscience & ML papers.

```python
plt.figure(figsize=(9, 4))

sns.violinplot(
    data=data,
    y="Group",
    x="Value",
    inner=None,
    cut=0,
)

sns.boxplot(
    data=data,
    y="Group",
    x="Value",
    width=0.2,
    showfliers=False,
)

sns.stripplot(
    data=data,
    y="Group",
    x="Value",
    jitter=0.25,
    alpha=0.4,
)

plt.title("Horizontal Raincloud Plot")
plt.show()
```

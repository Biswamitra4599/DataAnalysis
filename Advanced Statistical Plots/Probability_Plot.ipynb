## Normal Probability Plot (Q–Q Plot) — Standard
🔹 Idea

Compare sample quantiles vs theoretical normal quantiles.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Sample data
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=500)

plt.figure(figsize=(6, 6))
stats.probplot(data, dist="norm", plot=plt)
plt.title("Normal Probability Plot (Q–Q Plot)")
plt.show()
```

## Custom Q–Q Plot (Manual & Transparent)

Useful if you want full control.

```python
sorted_data = np.sort(data)
n = len(sorted_data)

# Theoretical quantiles
theoretical_q = stats.norm.ppf((np.arange(1, n + 1) - 0.5) / n)

plt.figure(figsize=(6, 6))
plt.scatter(theoretical_q, sorted_data, alpha=0.6)
plt.plot(theoretical_q, theoretical_q, linestyle='--')
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.title("Q–Q Plot (Manual Construction)")
plt.show()
```

## Probability Plot for Any Distribution
🔹 Example: Exponential Distribution

```python
exp_data = np.random.exponential(scale=1, size=500)

plt.figure(figsize=(6, 6))
stats.probplot(exp_data, dist="expon", plot=plt)
plt.title("Exponential Probability Plot")
plt.show()
```

## P–P Plot (Probability–Probability Plot)
🔹 Idea

Compare empirical CDF vs theoretical CDF.

```python
sorted_data = np.sort(data)
empirical_cdf = np.arange(1, len(data) + 1) / len(data)
theoretical_cdf = stats.norm.cdf(sorted_data)

plt.figure(figsize=(6, 6))
plt.scatter(theoretical_cdf, empirical_cdf, alpha=0.6)
plt.plot([0, 1], [0, 1], linestyle='--')
plt.xlabel("Theoretical CDF")
plt.ylabel("Empirical CDF")
plt.title("P–P Plot")
plt.show()
```

## Probability Plot with Confidence Bands (Advanced)

```python
osm, osr = stats.probplot(data, dist="norm", fit=False)

slope, intercept, r = stats.probplot(data, dist="norm")[1]

plt.figure(figsize=(6, 6))
plt.scatter(osm, osr, alpha=0.5)
plt.plot(osm, slope * osm + intercept, linewidth=2)
plt.title(f"Q–Q Plot (R² = {r**2:.3f})")
plt.show()
```

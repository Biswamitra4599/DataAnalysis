## Basic Rank Plot

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.lognormal(mean=1, sigma=1, size=300)

# Sort in descending order
data_sorted = np.sort(data)[::-1]
ranks = np.arange(1, len(data_sorted) + 1)

plt.figure(figsize=(7, 5))
plt.plot(ranks, data_sorted, "o-", alpha=0.7)

plt.xlabel("Rank")
plt.ylabel("Value")
plt.title("Basic Rank Plot")
plt.grid(alpha=0.3)
plt.show()
```

## Log–Log Rank Plot (Zipf / Power-Law Analysis)

Most rank–frequency data forms a straight line on log-log scales.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.zipf(a=2, size=500)

data_sorted = np.sort(data)[::-1]
ranks = np.arange(1, len(data_sorted) + 1)

plt.figure(figsize=(7, 5))
plt.loglog(ranks, data_sorted, "o", alpha=0.6)

plt.xlabel("Rank (log)")
plt.ylabel("Value (log)")
plt.title("Log–Log Rank Plot (Zipf Style)")
plt.grid(alpha=0.3, which="both")
plt.show()
```

## Rank Plot With Theoretical Power-Law Fit

Useful for verifying if the data follows a heavy-tail distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

np.random.seed(42)
data = np.random.zipf(2, size=500)
data_sorted = np.sort(data)[::-1]
ranks = np.arange(1, len(data_sorted) + 1)

# Fit power law y = a * x^(-b)
def power_law(x, a, b):
    return a * np.power(x, -b)

params, _ = curve_fit(power_law, ranks, data_sorted)

plt.figure(figsize=(7, 5))
plt.loglog(ranks, data_sorted, "o", label="Data", alpha=0.6)
plt.loglog(ranks, power_law(ranks, *params), "r--", lw=2,
           label=f"Power-Law Fit: y = {params[0]:.2f} * x^(-{params[1]:.2f})")

plt.xlabel("Rank (log)")
plt.ylabel("Value (log)")
plt.title("Rank Plot with Power-Law Fit")
plt.legend()
plt.grid(alpha=0.3, which="both")
plt.show()
```

## Ranked Scatter Plot With Labels (Top-K)

Useful when visualizing top-ranked items like words, cities, companies, etc.

```python
import numpy as np
import matplotlib.pyplot as plt

names = np.array(["A","B","C","D","E","F","G","H","I","J"])
values = np.array([80, 65, 60, 50, 48, 47, 30, 20, 10, 5])

data_sorted = np.sort(values)[::-1]
sorted_names = names[np.argsort(values)[::-1]]
ranks = np.arange(1, len(values) + 1)

plt.figure(figsize=(7, 5))
plt.scatter(ranks, data_sorted, s=80, color="royalblue")

for r, val, label in zip(ranks, data_sorted, sorted_names):
    plt.text(r, val + 1, label, ha="center")

plt.xlabel("Rank")
plt.ylabel("Value")
plt.title("Ranked Scatter Plot with Labels")
plt.grid(alpha=0.3)
plt.show()
```

## Rank Plots for Multiple Columns

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

df = pd.DataFrame({
    "Feature1": np.random.lognormal(mean=1, sigma=1, size=300),
    "Feature2": np.random.exponential(scale=2, size=300),
    "Feature3": np.random.uniform(1, 10, size=300)
})

plt.figure(figsize=(8, 6))

for col in df.columns:
    data_sorted = np.sort(df[col])[::-1]
    ranks = np.arange(1, len(data_sorted)+1)
    plt.plot(ranks, data_sorted, label=col)

plt.xlabel("Rank")
plt.ylabel("Value")
plt.title("Rank Plots for Multiple Features")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Professional Publication-Ready Rank Plot (Styled)

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")

np.random.seed(42)
data = np.random.lognormal(mean=1, sigma=1.1, size=500)

data_sorted = np.sort(data)[::-1]
ranks = np.arange(1, len(data_sorted) + 1)

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(ranks, data_sorted, marker="o", markersize=4,
        alpha=0.7, linewidth=1.5, color="navy")

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_title("Professional Rank Plot (Log–Log)", fontsize=16, weight="bold")
ax.set_xlabel("Rank (log scale)")
ax.set_ylabel("Value (log scale)")

ax.grid(alpha=0.3, which="both")
plt.show()
```

## Basic Violin + Swarm Plot (Seaborn)
🔹 Structure

Violin → density

Swarm → exact data points (no overlap)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample data
np.random.seed(42)
df = pd.DataFrame({
    "Value": np.concatenate([
        np.random.normal(0, 1, 200),
        np.random.normal(2, 1.2, 200)
    ]),
    "Group": ["A"] * 200 + ["B"] * 200
})

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="Group",
    y="Value",
    inner=None,
    cut=0,
)

sns.swarmplot(
    data=df,
    x="Group",
    y="Value",
    size=4,
)

plt.title("Violin–Swarm Hybrid Plot")
plt.show()
```

## Violin + Jittered Strip (Scales Better)

Swarm plots can struggle with large data. Use stripplot instead.

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="Group",
    y="Value",
    inner=None,
    cut=0,
)

sns.stripplot(
    data=df,
    x="Group",
    y="Value",
    jitter=0.25,
    alpha=0.5,
    size=3,
)

plt.title("Violin + Jittered Points")
plt.show()
```

## Half-Violin + Swarm (Modern Look)

A popular paper-style variant.

```python
plt.figure(figsize=(8, 5))

ax = sns.violinplot(
    data=df,
    x="Group",
    y="Value",
    inner=None,
    cut=0,
)

# Convert to half violins
for violin in ax.collections:
    violin.set_clip_path(
        plt.Rectangle(
            (violin.get_paths()[0].vertices[:, 0].mean(), -10),
            10, 20,
            transform=ax.transData
        )
    )

sns.swarmplot(
    data=df,
    x="Group",
    y="Value",
    size=4,
)

plt.title("Half-Violin Swarm Plot")
plt.show()
```

## Horizontal Violin–Swarm Plot

Better for long labels & distributions.

```python
plt.figure(figsize=(9, 4))

sns.violinplot(
    data=df,
    y="Group",
    x="Value",
    inner=None,
    cut=0,
)

sns.swarmplot(
    data=df,
    y="Group",
    x="Value",
    size=4,
)

plt.title("Horizontal Violin–Swarm Plot")
plt.show()
```

## Multi-Class Violin–Swarm (Hue-Based)

```python
df_multi = df.copy()
df_multi["Condition"] = np.random.choice(["Control", "Treatment"], size=len(df))

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df_multi,
    x="Group",
    y="Value",
    hue="Condition",
    inner=None,
    split=True,
    cut=0,
)

sns.swarmplot(
    data=df_multi,
    x="Group",
    y="Value",
    hue="Condition",
    dodge=True,
    size=3,
)

plt.title("Violin–Swarm with Multiple Conditions")
plt.legend(bbox_to_anchor=(1.05, 1))
plt.show()
```

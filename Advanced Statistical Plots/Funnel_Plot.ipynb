## Basic Funnel Plot

Used in meta-analysis to visually inspect publication bias.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Simulated effect sizes and standard errors
effect = np.random.normal(0, 0.3, 30)
se = np.random.uniform(0.05, 0.25, 30)

plt.figure(figsize=(7, 6))
plt.scatter(effect, se, alpha=0.7)

plt.gca().invert_yaxis()  # standard funnel plot convention

plt.xlabel("Effect Size")
plt.ylabel("Standard Error")
plt.title("Basic Funnel Plot")
plt.grid(alpha=0.3)
plt.show()
```

## Funnel Plot with 95% Confidence Region

This creates the characteristic inverted funnel shape.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

effect = np.random.normal(0, 0.25, 35)
se = np.random.uniform(0.05, 0.30, 35)

mean_effect = np.mean(effect)

# Funnel boundaries at ±1.96 * SE
se_range = np.linspace(min(se), max(se), 100)
upper = mean_effect + 1.96 * se_range
lower = mean_effect - 1.96 * se_range

plt.figure(figsize=(7, 6))
plt.scatter(effect, se, alpha=0.7)

plt.plot(upper, se_range, 'red', linestyle='--')
plt.plot(lower, se_range, 'red', linestyle='--')

plt.gca().invert_yaxis()
plt.xlabel("Effect Size")
plt.ylabel("Standard Error")
plt.title("Funnel Plot with 95% Confidence Region")
plt.grid(alpha=0.3)
plt.show()
```

## Funnel Plot with Egger Regression Line

Egger’s test detects asymmetry, indicating publication bias.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(42)

effect = np.random.normal(0, 0.3, 40)
se = np.random.uniform(0.05, 0.25, 40)
precision = 1 / se

# Egger regression: effect/SE ~ precision
X = precision.reshape(-1, 1)
y = effect / se

reg = LinearRegression().fit(X, y)
line = reg.predict(X)

plt.figure(figsize=(7, 6))
plt.scatter(effect, se, alpha=0.7)

# Funnel lines
mean_effect = np.mean(effect)
se_range = np.linspace(min(se), max(se), 100)
plt.plot(mean_effect + 1.96*se_range, se_range, 'gray', linestyle='--')
plt.plot(mean_effect - 1.96*se_range, se_range, 'gray', linestyle='--')

# Egger asymmetry line (transformed to original scale)
egger_x = np.linspace(min(effect), max(effect), 100)
egger_y = (reg.intercept_ + reg.coef_[0] * (1/se_range))/se_range

plt.plot(egger_x, np.interp(egger_x, effect, se), 'r--', label="Egger Trend (Asymmetry)")

plt.gca().invert_yaxis()
plt.xlabel("Effect Size")
plt.ylabel("Standard Error")
plt.title("Funnel Plot with Egger Regression (Bias Detection)")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Funnel Plot with Bubble Sizes (Weight = 1/SE²)

Larger studies → smaller SE → larger bubbles.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
effect = np.random.normal(0, 0.2, 40)
se = np.random.uniform(0.05, 0.30, 40)
weights = 1 / (se ** 2)

plt.figure(figsize=(7, 6))
plt.scatter(effect, se, s=weights * 5, alpha=0.6, color='royalblue')

# Funnel boundaries
mean_effect = np.mean(effect)
se_range = np.linspace(min(se), max(se), 100)
plt.plot(mean_effect + 1.96*se_range, se_range,'r--')
plt.plot(mean_effect - 1.96*se_range, se_range,'r--')

plt.gca().invert_yaxis()
plt.xlabel("Effect Size")
plt.ylabel("Standard Error")
plt.title("Bubble Funnel Plot (Weighted by Study Precision)")
plt.grid(alpha=0.3)
plt.show()
```

## Trim-and-Fill Funnel Plot (Advanced Meta-Analysis)

This technique attempts to correct publication bias by filling missing studies.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
effect = np.random.normal(0, 0.25, 30)
se = np.random.uniform(0.05, 0.25, 30)

# Fake missing studies (mirrored for demonstration)
missing_effect = -effect + 2 * np.mean(effect)
missing_se = se

plt.figure(figsize=(7, 6))

# Actual studies
plt.scatter(effect, se, color="navy", alpha=0.7, label="Observed")

# Filled studies
plt.scatter(missing_effect, missing_se, color="orange", alpha=0.6, label="Filled (Trim-and-Fill)")

# Funnel boundaries
mean_effect = np.mean(effect)
se_range = np.linspace(min(se), max(se), 100)
plt.plot(mean_effect + 1.96*se_range, se_range, 'gray', linestyle='--')
plt.plot(mean_effect - 1.96*se_range, se_range, 'gray', linestyle='--')

plt.gca().invert_yaxis()
plt.xlabel("Effect Size")
plt.ylabel("Standard Error")
plt.title("Trim-and-Fill Funnel Plot")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Publication-Ready Funnel Plot (Beautiful + Clean)

```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")

np.random.seed(42)
effect = np.random.normal(0.05, 0.2, 40)
se = np.random.uniform(0.05, 0.20, 40)

mean_effect = np.mean(effect)
se_range = np.linspace(min(se), max(se), 200)

fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(effect, se, color='navy', alpha=0.8, edgecolor='white', s=70)

ax.plot(mean_effect + 1.96*se_range, se_range, 'r--', lw=2)
ax.plot(mean_effect - 1.96*se_range, se_range, 'r--', lw=2)

ax.set_title("Publication-Ready Funnel Plot", fontsize=16, fontweight='bold')
ax.set_xlabel("Effect Size")
ax.set_ylabel("Standard Error")
ax.invert_yaxis()
ax.grid(alpha=0.25)

plt.show()
```

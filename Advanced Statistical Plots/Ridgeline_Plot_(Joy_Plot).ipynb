## Ridgeline Plot

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

np.random.seed(42)

# Create multiple distributions
groups = {
    "Group A": np.random.normal(0, 1, 400),
    "Group B": np.random.normal(1, 1.2, 400),
    "Group C": np.random.normal(2, 0.8, 400),
    "Group D": np.random.normal(3, 1.5, 400),
}

fig, ax = plt.subplots(figsize=(8, 6))

y_offset = 0
for label, data in groups.items():
    kde = gaussian_kde(data)
    x = np.linspace(min(data)-2, max(data)+2, 300)
    y = kde(x)
    
    ax.fill_between(x, y + y_offset, y_offset, alpha=0.6)
    ax.text(min(x), y_offset + max(y)*0.8, label, fontsize=12)
    
    y_offset += 1.2  # spacing

ax.set_yticks([])
ax.set_xlabel("Value")
ax.set_title("Ridgeline (Joy) Plot - Matplotlib")
plt.show()
```

## Ridgeline Plot Using JoyPy (The Standard Way)

JoyPy makes ridgelines very easy.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "A": np.random.normal(0, 1, 500),
    "B": np.random.normal(1, 1.1, 500),
    "C": np.random.normal(2, 0.8, 500),
    "D": np.random.normal(3, 1.4, 500),
})

plt.figure(figsize=(10, 6))

for i, column in enumerate(df.columns):
    sns.kdeplot(df[column], fill=True)
    plt.text(df[column].mean(), 0.15 + i * 0.05, column, fontsize=12)

plt.title("Ridgeline Plot (Manual Seaborn Version)")
plt.show()
```

## Ridgeline Plot for Many Distributions

```python
import numpy as np
import pandas as pd
import joypy
import matplotlib.pyplot as plt

np.random.seed(42)

cols = [f"Dist_{i}" for i in range(12)]
df = pd.DataFrame({c: np.random.normal(i, 1, 400) for i, c in enumerate(cols)})

joypy.joyplot(
    df,
    figsize=(10, 8),
    colormap=plt.cm.coolwarm,
    fade=True
)

plt.title("Ridgeline Plot for Many Distributions")
plt.show()
```

## Custom Color Gradients + Transparency

```python
import numpy as np
import pandas as pd
import joypy
import matplotlib.pyplot as plt

np.random.seed(42)

df = pd.DataFrame({
    "Low": np.random.normal(0, 1, 500),
    "Mid": np.random.normal(1, 1.2, 500),
    "High": np.random.normal(2, 1.4, 500),
})

fig, axes = joypy.joyplot(
    df,
    figsize=(9, 6),
    colormap=plt.cm.plasma,   # ✅ FIXED
    kind="kde",
    alpha=0.5,
    linewidth=1.2,
    overlap=0.7
)

plt.title("Ridgeline Plot with Plasma Gradient", fontsize=16)
plt.show()
```

## Ridgeline with KDE Smoothing (Manual, Matplotlib)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

np.random.seed(42)

datasets = [
    np.random.normal(0, 1, 300),
    np.random.normal(1, 1.3, 300),
    np.random.normal(2, 0.9, 300),
    np.random.normal(3, 1.6, 300),
]

fig, ax = plt.subplots(figsize=(8, 6))

offset = 0
colors = ["#4C72B0", "#55A868", "#C44E52", "#8172B2"]

for idx, data in enumerate(datasets):
    kde = gaussian_kde(data)
    x = np.linspace(min(data)-2, max(data)+2, 400)
    y = kde(x)

    ax.plot(x, y + offset, color=colors[idx], lw=2)
    ax.fill_between(x, offset, y + offset, color=colors[idx], alpha=0.5)
    
    ax.text(min(x), offset + max(y)*0.8, f"Dataset {idx+1}", fontsize=12)
    
    offset += 1.1

ax.set_yticks([])
ax.set_title("Smoothed KDE Ridgeline Plot (Matplotlib)")
ax.set_xlabel("Value")
plt.show()

```

## Publication-Ready Ridgeline Plot (Beautiful & Clean)

```python
import numpy as np
import pandas as pd
import joypy
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")
np.random.seed(42)

df = pd.DataFrame({
    "2020": np.random.normal(0, 1, 700),
    "2021": np.random.normal(0.5, 1.1, 700),
    "2022": np.random.normal(1, 1.2, 700),
    "2023": np.random.normal(1.5, 1.3, 700),
    "2024": np.random.normal(2, 1.4, 700),
})

fig, axes = joypy.joyplot(
    df,
    figsize=(10, 7),
    colormap=plt.cm.inferno_r,
    fade=True,
    overlap=0.7,
    linewidth=1.3,
    kind="kde",
    alpha=0.8
)

plt.title("Publication-Ready Ridgeline Plot", fontsize=18, fontweight="bold")
plt.show()
```

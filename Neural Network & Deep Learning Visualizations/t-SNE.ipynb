## t-SNE

### Basic t-SNE Plot (scikit-learn)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.manifold import TSNE

# Example embeddings
data = load_iris()
X = data.data  # embeddings
y = data.target  # labels

# t-SNE
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X)

# Plot
plt.figure(figsize=(7, 5))
for label in np.unique(y):
    plt.scatter(X_tsne[y==label, 0], X_tsne[y==label, 1], label=f"Class {label}", alpha=0.7)

plt.legend()
plt.title("t-SNE Plot of Embeddings")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.show()
```

### Colored t-SNE with Seaborn

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "tsne-1": X_tsne[:, 0],
    "tsne-2": X_tsne[:, 1],
    "label": y
})

plt.figure(figsize=(7, 5))
sns.scatterplot(
    x="tsne-1", y="tsne-2",
    hue="label",
    palette=sns.color_palette("tab10", n_colors=len(np.unique(y))),
    data=df,
    alpha=0.8
)
plt.title("t-SNE Plot (Seaborn)")
plt.show()
```

### 3D t-SNE Plot

```python
from mpl_toolkits.mplot3d import Axes3D

tsne = TSNE(n_components=3, random_state=42)
X_tsne_3d = tsne.fit_transform(X)

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
for label in np.unique(y):
    ax.scatter(X_tsne_3d[y==label, 0], X_tsne_3d[y==label, 1], X_tsne_3d[y==label, 2], label=f"Class {label}")
ax.set_title("3D t-SNE Plot")
plt.legend()
plt.show()
```

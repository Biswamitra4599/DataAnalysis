## Basic UMAP Plot

### Basic UMAP Plot (2D)


```python
import umap
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Example embeddings
data = load_iris()
X = data.data  # embeddings
y = data.target  # labels

# UMAP
reducer = umap.UMAP(n_components=2, random_state=42)
X_umap = reducer.fit_transform(X)

# Plot
plt.figure(figsize=(7,5))
for label in np.unique(y):
    plt.scatter(X_umap[y==label, 0], X_umap[y==label, 1], label=f"Class {label}", alpha=0.7)

plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.title("UMAP Plot of Embeddings")
plt.legend()
plt.show()
```

### UMAP with Seaborn

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "UMAP1": X_umap[:, 0],
    "UMAP2": X_umap[:, 1],
    "label": y
})

plt.figure(figsize=(7,5))
sns.scatterplot(
    x="UMAP1", y="UMAP2",
    hue="label",
    palette=sns.color_palette("tab10", n_colors=len(np.unique(y))),
    data=df,
    alpha=0.8
)
plt.title("UMAP Plot (Seaborn)")
plt.show()
```
